```python
def plot_sem(time, trace, ax, ii, label):

    colors = sns.color_palette('Set1',3)
    mean_trace = trace.mean(axis=0)
    sem_trace = trace.std(axis=0)/np.sqrt(trace.shape[0])
    ax.plot(time, mean_trace, color=colors[ii], label=label)
    ax.fill_between(time,
                    mean_trace - sem_trace,
                    mean_trace + sem_trace,
                    color=colors[ii],
                    alpha=0.3)
```

```python
def epoch_permutation(animal,date,n_perm = 1000, query='trialerror<2'):

    nwbfile = load_data.load_nwbfile(animal, date)
    trial_df = nwbfile.intervals['trials'].to_dataframe().query(query)
    node_df = nwbfile.intervals['node_on'].to_dataframe().query(query).query('step==0')

    start_times = trial_df.start_time.values
    node_ons = node_df.t_on.values-start_times
    n_neurons = nwbfile.acquisition['neural'].data.shape[1]
    
    node_FR = np.zeros((len(trial_df)-1,n_neurons))
    null_FR = np.zeros((n_perm,len(trial_df)-1,n_neurons))
    
    for trial in range(len(trial_df)-1):
    
        t_start = int(start_times[trial])
        t_end = int(start_times[trial+1])
        trial_spikes = nwbfile.acquisition['neural'].data[t_start:t_end,:]
        trial_length = len(trial_spikes)
        
        first_node_time = np.arange(node_ons[trial],node_ons[trial]+500).astype(int)
        node_FR[trial] = trial_spikes[first_node_time,:].sum(axis=0)
        
        trial_time = np.arange(trial_length)
        rest = trial_time[~np.isin(trial_time, np.arange(node_ons[trial]-500,node_ons[trial]+500)) & (trial_time < (trial_length-500))]
        
        for n in range(n_perm):
            null_idx = int(np.random.choice(rest))
            null_time = np.arange(null_idx,null_idx+500).astype(int)
            null_FR[n,trial,:] = trial_spikes[null_time,:].sum(axis=0)

    unitNames = nwbfile.units.to_dataframe()
    unitNames['node_FR'] = node_FR.mean(axis=0)
    unitNames['null_FR'] = list(null_FR.mean(axis=1).T)
    unitNames['animal'] = animal
    unitNames['session'] = session

    return unitNames
```

```python
def get_noramlized_epoch_mean(nwbfile, neural, trialtype='nav', slide=10):

    unit_params = {"drift": 2, "min_fr": 1}
    
    if trialtype=='nav':

        query='trialerror==0'
    
        trial_df = nwbfile.intervals['trials'].to_dataframe().query(query)
        node_df = nwbfile.intervals['node_on'].to_dataframe().query(query)
        node_df = node_df[~np.isin(node_df.node.values,[1,7,8,14])]
        choice_df = nwbfile.intervals['action_on'].to_dataframe().query(query)
        choice_df = choice_df[~np.isin(choice_df.node.values,[1,7,8,14])]
        reward_df = nwbfile.intervals['reward_on'].to_dataframe().query(query)
        
        epochs = {
            "trial_on": trial_df.start_time.values.astype(np.int64),
            "first_node_on": node_df.query('step==0').t_on.values.astype(np.int64),
            "first_action_on": choice_df.query('step==0').t_on.values.astype(np.int64),
            "next_action_on": choice_df.query('step>0').t_on.values.astype(np.int64),
            "next_node_on": node_df.query('step>0').t_on.values.astype(np.int64),
            "reward_on": reward_df.t_on.values.astype(np.int64),
        }
        
        epoch_windows = {
            "trial_on":      (-500, 500),
            "first_node_on": (-500, 500),
            "first_action_on": (-500, 500),
            "next_action_on": (-500, 500),
            "next_node_on":  (-500, 500),
            "reward_on":     (-500, 2500),
        }


    if trialtype=='flash':
        
        query='trialerror==1'
        
        trial_df = nwbfile.intervals['trials'].to_dataframe().query(query)
        node_df = nwbfile.intervals['node_on'].to_dataframe().query(query)
        choice_df = nwbfile.intervals['action_on'].to_dataframe().query(query)
        reward_df = nwbfile.intervals['reward_on'].to_dataframe().query(query)
            
        
        epochs = {
            "trial_on": trial_df.start_time.values.astype(np.int64),
            "first_node_on": node_df.query('step==0').t_on.values.astype(np.int64),
            "first_action_on": choice_df.query('step==0').t_on.values.astype(np.int64),
            "reward_on": reward_df.t_on.values.astype(np.int64),
        }
        
        epoch_windows = {
            "trial_on":      (-500, 500),
            "first_node_on": (-500, 500),
            "first_action_on": (-500, 500),
            "reward_on":     (-500, 2500),
        }
    
    unitNames = nwbfile.units.to_dataframe()
    unit_idx =  np.where((unitNames.group == "good") & (unitNames.drift <= unit_params["drift"]) & (unitNames.fr >= unit_params["min_fr"]))[0]
    unitNames = unitNames.iloc[unit_idx, :].reset_index(drop=True)
    ncell  = len(unitNames)
    
    # ---- bin spikes per epoch ----
    
    for epoch, times in epochs.items():
    
        t0, t1 = epoch_windows[epoch]
    
        epoch_spikes = np.empty((len(times), int((t1-t0)/10), ncell), dtype=float)
    
        for i, t in enumerate(times):
            epoch_spikes[i] = neural[t + t0 : t + t1 : slide]
    
        unitNames[epoch] = list(epoch_spikes.mean(axis=0).T)
        

    return unitNames
```

```python
def trial_permutation(animal,date,n_perm = 1000, query='trialerror<2'):

    nwbfile = load_data.load_nwbfile(animal, date)
    unitNames = nwbfile.units.to_dataframe()
    unit_idx = np.where((unitNames.group == "good") &
        (unitNames.drift <= 2) &
        (unitNames.fr >= 1))[0]
    unitNames = unitNames.iloc[unit_idx]

    
    trial_df = nwbfile.intervals['trials'].to_dataframe().query(query)
    trial_cc = pd.DataFrame(
    trial_df.drop_duplicates('trial')
      .groupby('block')
      .cumcount()
      .rename('trial_in_block')
    )
    
    trial_cc['trial'] = trial_df.trial.unique()
    # trial_df = trial_df.merge(trial_cc, on='trial')
    
    # trial_df["mean_FR"] = trial_df.apply(
    #     lambda row: row.timeseries[0].data[:, unit_idx].mean(axis=0)*1000,
    #     axis=1
    # )

    # get mean FR from the first node on
    trial_df = nwbfile.intervals['action_on'].to_dataframe().query('(trialerror<2)&(step==0)')
    trial_df = trial_df.merge(trial_cc, on='trial')
    trial_df["mean_FR"] = trial_df.apply(
        lambda row: row.timeseries[0].data[500:, unit_idx].mean(axis=0)*1000,
        axis=1
    )
    
    blocknumber = trial_df.block.unique()[trial_df.groupby('block').size().values>5]
    ncell=len(unitNames)
    null = np.zeros((5,n_perm,len(blocknumber),ncell))
    probe_FR = np.zeros((5,len(blocknumber),ncell))

    
    for probe_trial in range(5):
        
        for ii, block in enumerate(blocknumber):
            
            block_trial_FR = np.stack(trial_df.query('block==@block').mean_FR.values)
            ntrial = block_trial_FR.shape[0]
            null_idx = np.arange(ntrial)[np.arange(ntrial)!=probe_trial]
            probe_FR[probe_trial,ii,:] = block_trial_FR[probe_trial,:]
        
            for nn in range(n_perm):
                idx = int(np.random.choice(null_idx))
                null[probe_trial,nn,ii,:] = block_trial_FR[idx,:]

    
    unitNames['trial_FR'] = list(probe_FR.mean(axis=1).T)
    unitNames['null_FR'] = list(null.mean(axis=2).T)
    unitNames['animal'] = animal
    unitNames['session'] = session

    return unitNames
```

```python
def plot_heatmap(concat_sdf_norm,sort_idx,region,trialtype='nav'):

    epoch_windows = {
        "trial_on":      (-500, 500),
        "first_node_on": (-500, 500),
        "first_action_on": (-500, 500),
        "next_action_on": (-500, 500),
        "next_node_on":  (-500, 500),
        "reward_on":     (-500, 2500),
    }
    
    # Example: durations for relative widths
    durations = np.array([
        epoch_windows[e][1] - epoch_windows[e][0]
        for e in epoch_windows.keys()
    ])
    width_ratios = durations / durations.min()  # relative widths
    
    # Compute shared vmin/vmax across all data
    vmin = -1.5 #np.nanmin(concat_sdf_norm)
    vmax = 1.5 #np.nanmax(concat_sdf_norm)
    
    plt.close('all')
    fig, axes = plt.subplots(
        1, len(epoch_windows),
        figsize=(18, 6),
        sharey=True,
        gridspec_kw={"width_ratios": width_ratios}
    )
    
    time_idx = 0
    for jj, key in enumerate(epoch_windows.keys()):

        if key in concat_sdf_norm.columns:
            sns.heatmap(
                np.vstack(concat_sdf_norm[key].values)[sort_idx,:],
                ax=axes[jj],
                cmap='bwr',
                cbar=(jj==len(epoch_windows)-1),  # show colorbar only on last
                vmin=vmin,
                vmax=vmax
            )
            
            # mark event
            axes[jj].axvline(-epoch_windows[key][0]/10, color='lightgrey')
            axes[jj].set_title(key)
            axes[jj].set_xticks(np.arange(25,int(np.diff(epoch_windows[key])[0]/10),25))
            axes[jj].set_xticklabels(np.arange(epoch_windows[key][0]+250, epoch_windows[key][1], 250),rotation=45)
    
            if jj>0:
                axes[jj].set_yticks([])
            else:
                axes[jj].set_yticks(np.arange(100,len(concat_sdf_norm),100))
                axes[jj].set_yticklabels(np.arange(100, len(concat_sdf_norm), 100))
                axes[jj].tick_params(axis='y', labelleft=True)


    axes[0].set_ylabel("Neurons", labelpad=40)

    # fig.text(0.02, 0.5, "Neurons", va='center', rotation='vertical')

    plt.tight_layout(rect=[0,0,1,0.9])
    # plt.suptitle(f'{animal} {region}, {trialtype} epoch tuning')
    # plt.savefig(f'D:/Epoch/{trialtype} epoch tuning heatmap noramlized FR, {animal} {region}')
    plt.suptitle(f'{region}, {trialtype} epoch tuning')
    # plt.savefig(f'D:/Epoch/{trialtype} epoch tuning heatmap noramlized FR, {region}')
```

```python
import pandas as pd
import scipy.stats as stats

# Create the contingency table
contingency_table = pd.crosstab(epoch_tuning['epoch_tuning'], epoch_tuning['value'])

# Perform the Chi-Square test of independence
chi2_stat, p_val, dof, expected = stats.chi2_contingency(contingency_table)

# Output the results
print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-value: {p_val}")
print(f"Degrees of Freedom: {dof}")
print(f"Expected Frequencies Table:\n{expected}")

# Interpret the result
if p_val < 0.05:
    print("Reject the null hypothesis: The variables are dependent.")
else:
    print("Fail to reject the null hypothesis: The variables are independent.")
```

```python
def raster_sdf_aligned(
    timestamps,
    neuron_spikes,
    neuron_sdf,
    event_times,
    window
):
    """
    event_times: array of event timestamps (ms)
    window: (t_start, t_end) relative to event
    """
    ncell = neuron_spikes.shape[1]
    aligned_spikes = np.zeros((len(event_times),window[1]-window[0],ncell), dtype=int)
    aligned_sdf = np.zeros((len(event_times),window[1]-window[0],ncell), dtype=int)

    for i, t0 in enumerate(event_times):
        t_start = t0 + window[0]
        t_end   = t0 + window[1]

        idx = (timestamps >= t_start) & (timestamps < t_end)
        aligned_spikes[i] = neuron_spikes[idx]
        aligned_sdf[i] = neuron_sdf[idx]

    return aligned_spikes, aligned_sdf
```

```python
def get_aligned_sdf():
    
    aligned_spikes = {}
    aligned_sdf = {}
    
    for epoch_name, event_times in epochs.items():
        
        win = epoch_windows[epoch_name]
            
        aligned_spikes[epoch_name], aligned_sdf[epoch_name] = raster_sdf_aligned(
            timestamps,
            neuron_spikes,
            neuron_sdf,
            event_times,
            win
        )
    
    save_path = "D:/Epoch/HPC/flash_epoch_sdf_Session_%s.pkl" % date
    
    # Save both objects together in a dictionary
    with open(save_path, "wb") as f:
        pickle.dump({
            "aligned_spikes": aligned_spikes,
            "aligned_sdf": aligned_sdf
        }, f)

    return aligned_spikes, aligned_sdf
```

```python
def get_epoch_meanFR(df, next_df=[]):
        
    concat_sdf_windowed = []
    trialnumbers = df.trial_in_block.values
        
    for key in aligned_spikes.keys():

        epoch_spikes = aligned_spikes[key]

        if key[:4]=='next':
            trialnumbers = next_df.trial_in_block.values
        else:
            trialnumbers = df.trial_in_block.values
        
        n_windows = epoch_spikes.shape[1] // window_size
        n_neurons = epoch_spikes.shape[2]
        windowed_spikes = np.zeros((n_windows,len(np.unique(trialnumbers)),n_neurons))

        for ii, val in enumerate(np.unique(trialnumbers)):
            
            # mean across trials
            sdf_mean = epoch_spikes[trialnumbers==val,:,:].mean(axis=0)*1000  # shape: (time, n_neurons)
            
            # take full windows and compute mean along each window
            windowed_spikes[:,ii,:] = sdf_mean[:n_windows*window_size].reshape(n_windows, window_size, -1).mean(axis=1)
            
        concat_sdf_windowed.append(windowed_spikes)

    # concat_sdf_windowed = np.vstack(concat_sdf_windowed)

    # return np.vstack(concat_sdf_windowed).mean(axis=0)
    return concat_sdf_windowed
```

```python
def get_concat_sdf():
    
    window_size = 300  # rows per window
    
    concat_sdf_windowed = []
    
    for key in aligned_sdf.keys():
        # mean across trials
        sdf_mean = aligned_sdf[key].mean(axis=0)  # shape: (time, n_neurons)
        
        # compute number of full windows
        n_windows = sdf_mean.shape[0] // window_size
        
        # take full windows and compute mean along each window
        sdf_windowed = sdf_mean[:n_windows*window_size].reshape(n_windows, window_size, -1).mean(axis=1)
        
        concat_sdf_windowed.append(sdf_windowed)
    
    # stack all keys
    concat_sdf_windowed = np.vstack(concat_sdf_windowed)
    
    concat_sdf = np.vstack([aligned_sdf[key].mean(axis=0) for key in aligned_spikes.keys()])
    
    row_mean = concat_sdf.mean(axis=0, keepdims=True)
    row_std = concat_sdf.std(axis=0, keepdims=True)
    
    concat_sdf_norm = (concat_sdf - row_mean) / row_std
    
    
    # concat_sdf: (630, 90)
    # row_min = concat_sdf.min(axis=0, keepdims=True)
    # row_max = concat_sdf.max(axis=0, keepdims=True)
    
    # concat_sdf_norm = (concat_sdf - row_min) / (row_max - row_min)
    
    sort_idx = np.argsort(np.argmax(concat_sdf_windowed,axis=0))

    return concat_sdf_norm, sort_idx
```

```python
def plot_heatmap(concat_sdf_norm,sort_idx,region):
    # Example: durations for relative widths
    durations = np.array([
        epoch_windows[e][1] - epoch_windows[e][0]
        for e in epoch_windows.keys()
    ])
    width_ratios = durations / durations.min()  # relative widths
    
    # Compute shared vmin/vmax across all data
    vmin = -2 #np.nanmin(concat_sdf_norm)
    vmax = 2 #np.nanmax(concat_sdf_norm)
    
    plt.close('all')
    fig, axes = plt.subplots(
        1, len(epoch_windows),
        figsize=(18, 6),
        sharey=True,
        gridspec_kw={"width_ratios": width_ratios}
    )
    
    time_idx = 0
    for jj, key in enumerate(aligned_sdf.keys()):
        dur = np.diff(epoch_windows[key])[0]
        
        sns.heatmap(
            concat_sdf_norm.T[sort_idx, time_idx:time_idx+dur],
            ax=axes[jj],
            cmap='bwr',
            cbar=(jj==len(epoch_windows)-1),  # show colorbar only on last
            vmin=vmin,
            vmax=vmax
        )
        
        # mark event
        axes[jj].axvline(-epoch_windows[key][0], color='w')
        axes[jj].set_title(key)
        time_idx += dur
    
    fig.text(0.02, 0.5, "Neurons", va='center', rotation='vertical')
    plt.tight_layout(rect=[0,0,1,0.9])
    plt.suptitle(f'{region}, trial progression heatmap std')
    # plt.savefig(f'D:/Epoch/{region}/flash trial progression heatmap std')
```

```python
def process_channel(channel):
    coeffs = np.zeros((n_trials, n_freqs, n_time), dtype=np.complex64)
    
    for trial in range(n_trials):
        _, coeffs[trial] = LFP.wavelet(
            aligned_LFP[trial, channel, :],
            freq_range
        )
    
    return channel, coeffs
```

```python
for date in ['032823','033123','032023','040423','040723']:
    
    epoch = 'node_on'
    window = [-500,1000]
    
    
    nwbfile = load_data.load_nwbfile('Bart',date) 
    df = nwbfile.intervals[epoch].to_dataframe()
    df = df[df.trialerror==0]
    df['graph_distance'] = np.clip(df['graph_distance'].values,0,3)
    df = graph.append_use_tele(df)
    
    aligned_LFP = load_data.epoch_aligned_lfp(date,df,window)  # this is already notch filtered
    
    freq_range = np.arange(2,30)
    
    time = np.arange(window[0], window[1])
    cmin = 0
    colormap = 'jet'
    
    n_channels = aligned_LFP.shape[1]
    n_trials = aligned_LFP.shape[0]
    n_freqs = len(freq_range)
    n_time = len(time)
    
    channels = np.arange(n_channels)
    
    parallel = Parallel(n_jobs=8, return_as='generator')
    
    
    # Create HDF5 file and dataset
    h5_path = f'D:/Spectrogram/Session{date}_{epoch}_wavelet.h5'
    
    with h5py.File(h5_path, 'w') as f:
        dset = f.create_dataset(
            'wavelet_coeffs',
            shape=(n_channels, n_trials, n_freqs, n_time),
            dtype=np.complex64,
            chunks=(1, n_trials, n_freqs, n_time),  # chunk by channel
            compression='gzip'
        )
    
        # Process + write one channel at a time
        results = parallel(
            delayed(process_channel)(ch) for ch in tqdm(channels)
        )
    
        for ch, coeffs in results:
            dset[ch, :, :, :] = coeffs
```

### Cross-spectrogram
1. use coefficients obtained from wavelet analysis to get cross-spectrogram for a pair of signals (typically between two regions)
2. since it is not normalized by amplitude, it reflects if two signals both have high power at a certain frequency
3. a good complement to coherence, since coherence can only be caluclated across trials, but cross-spectrogram can provide insight at single-trial resolution. 
4. Equation  
    SXY = xf*conj(yf)                       Cross spectrum XY

```python
ch1 = 45
ch2 = 127
time = np.arange(10000)
freq_range = np.arange(2,100)  

sig1 = LFP.notchfilter(lfpData[time,ch1])
sig2 = LFP.notchfilter(lfpData[time,ch2])

freqs, xf = LFP.wavelet(sig1, freq_range)
freqs, yf = LFP.wavelet(sig2, freq_range)

SXX = np.real(xf*np.conj(xf))
SYY = np.real(yf*np.conj(yf))
SXY = xf*np.conj(yf)  
cross_power = np.abs(SXY)**2
phase = np.angle(SXY)
```

```python
plt.close('all')
fig, ax = plt.subplots()
ax.plot(freqs,phase.mean(axis=1),color='k')

ax2 = ax.twinx()
ax2.plot(freqs,np.log(cross_power.mean(axis=1)),color='r')
ax.axhline(0)
ax.axvspan(6,20,color='y',alpha=0.1)
```

### Coherogram
1. typically only obtained from multiple trials, single-trial coherence will need multitapering
2. Equation  
    SXX = real( xf*conj(xf)                 Spectrum X  
    SYY = real( yf*conj(yf)                 Spectrum Y  
    SXY = xf*conj(yf)                       Cross spectrum XY  
    coherence = np.abs(SXY) / np.sqrt(SXX) / np.sqrt(SYY)

```python
import os
import pynwb
import h5py
from tqdm import tqdm
import pandas as pd
from pathlib import Path
import numpy as np
import seaborn as sns
import pywt
import matplotlib
import graph
import time
import pickle

import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib import gridspec

from scipy import signal
from scipy.fftpack import next_fast_len
from scipy.io import loadmat
from scipy.ndimage import gaussian_filter1d
from scipy.signal import filtfilt, hilbert, remez, sosfiltfilt, butter, find_peaks
# import pingouin as pg
from scipy.stats import fisher_exact


import load_data
import LFP
import single_neuron_vis
```

```python
ses_info = pd.DataFrame({
    'session': ['031723','032823','033123','032023','040423','040723','050824',
                '101124','100324','092124','092324','092524','100124','100924'],
    'animal': ['Bart']*7 + ['London']*7
})


action_all = pd.DataFrame()

for _, row in ses_info.iterrows():
    date = row['session']
    animal = row['animal']

    nwbfile = load_data.load_nwbfile(animal, date)

    action_df = (
        nwbfile.intervals['action_on']
        .to_dataframe()
        .query('trialerror==0')
        .copy()
    )

    # add metadata (very important)
    action_df['session'] = date
    action_df['animal'] = animal

    action_all = pd.concat([action_all,action_df])

action_all = action_all.drop(columns=['timeseries'])

first = action_all.query('step==0').copy()
first['trial_in_block'] = (
    first
    .groupby(['session', 'block'])
    .cumcount()
)

block_targets = (
    first
    .drop_duplicates(subset=['session', 'block'])
    [['session', 'block', 'target']]
)

block_targets['prev_block_target'] = (
    block_targets
    .groupby('session')['target']
    .shift(1)
)

first = first.merge(
    block_targets[['session', 'block', 'prev_block_target']],
    on=['session', 'block'],
    how='left'
)

first['optimality'] = first['graph_distance']/first['nsteps']
first['extra'] = first['nsteps']-first['graph_distance']
first['steps_saved']=first['spatial_distance']-first['graph_distance']
```

```python
block_transition_counts = first[(first.trial_in_block==0)&(first.animal==animal)].\
                          groupby(['prev_block_target','target']).size().reset_index(name='counts')
block_transition_counts = block_transition_counts.query('counts>10')
block_transitions = first.merge(block_transition_counts,how='inner', on=['prev_block_target','target'])
```

```python
# look at the behavior across block transition
# which transitions are harder

animal = 'Bart'
pivot = block_transitions[(block_transitions.trial_in_block==0)&(block_transitions.animal==animal)].pivot_table(
    index='prev_block_target',
    columns='target',
    values='extra',
    aggfunc='mean'
)

plt.figure(figsize=(5,4))

sns.heatmap(
    pivot,
    annot=True,
    fmt='.2f',
    cmap='Blues',
    cbar_kws={'label': 'Extra steps'},
    vmin = 0,
    vmax = 0.8,
)

plt.xlabel('Current target')
plt.ylabel('Previous block target')
plt.title(f'{animal}, Route optimality | block transition')
plt.tight_layout()
# plt.savefig(f'D:/Behavior/{animal}, Route optimality, block transition')
```

```python
animal='London'
df = block_transitions[
    (block_transitions.trial_in_block == 0) &
    (block_transitions.animal == animal) & (block_transitions.extra>0)
]
plt.close('all')
g = sns.FacetGrid(
    df,
    row='prev_block_target',
    col='target',
    margin_titles=True,
    height=2,
    aspect=1
)

g.map_dataframe(
    sns.histplot,
    x='extra',
    fill=True,
    common_norm=False,
    alpha=0.7,
    bins=np.arange(-0.5,15.5)
)

g.set_axis_labels("Extra steps", "Density")
g.fig.suptitle(f'{animal} | Extra steps by block transition', y=1.02)
plt.savefig(f'D:/Behavior/{animal}, Extra steps by block transition, distribution')
```

```python
# look at the behavior across block transition
# which transitions are harder

animal = 'London'
pivot = block_transitions[(block_transitions.trial_in_block==0)&(block_transitions.animal==animal)&(block_transitions.steps_saved==0)].pivot_table(
    index='prev_block_target',
    columns='target',
    values='extra',
    aggfunc='mean'
)

plt.figure(figsize=(5,4))

sns.heatmap(
    pivot,
    annot=True,
    fmt='.2f',
    cmap='Blues',
    cbar_kws={'label': 'Extra steps'},
    vmin = 0,
    vmax = 0.8,
)

plt.xlabel('Current target')
plt.ylabel('Previous block target')
plt.title(f'{animal}, Route optimality \nblock transition, no teleport')
plt.tight_layout()
plt.savefig(f'D:/Behavior/{animal}, Route optimality, block transition, no teleport')
```

```python
tele_trials = []
tele_perf = []
for ss in ['031723','032823','033123','032023','040423','040723','050824','101124','100324','092124','092324','092524','100124','100924']:
    session = first[first.session==ss]
    tele_trials.append(np.sum(session.steps_saved==1))
    tele_perf.append(np.sum((session.steps_saved==1) & (session.extra>0)))
```
